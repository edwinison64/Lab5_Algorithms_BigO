```python
#[CELL 3] LINEAR GROWTH
def linear_operation(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operation(n):5d}")
```

```python
# [CELL 4] BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count
    
for n in [10, 100, 1000]:
        print(
            f"n={n:4d} | O(1)=1 | "
            f"O(log n)={log_ops(n):2d} | "
            f"O(n)={n:4d} | O(n^2)={n*n:7d}"
        )
```

```python
# [CELL 5] SPACE COMPLEXITY 
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data)}")
```

```python
# [CELL 6] CASE ANALYSIS
def search_case(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i
        return len(data)

data = [11, 22, 33, 44, 55]

for case, target in {"Best": 11, "Average": 33, "Worst": 99}.items():
    print(f"{case} Case: {search_case(data, target)} comparisons")
  
```

```python
# [CELL 7] LINEAR SEARCH
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 8] BINARY SEARCH
def binary_search(data, target):
    low, high, comparisos = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisos += 1
        if data[mid] == target:
            return mid, comparisos
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisos

Index, comparisons = binary_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 9] SEARCH COMPARISON
data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1

li, lc = linear_search(data, target)
bi, bc = binary_search(data, target)

print("Target:", target)
print("Linear Search: Index =", li, ", Comparisons =", lc)
print("Binary Search: Index =", bi, ", Comparisons =", bc)
print("Reduction:", lc - bc)
```

```python
# [CELL 10] BUBBLE SORT
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - 1 - i):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1 
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100, (SEED_DIGIT * 11) % 100, (NAME_LENGTH * 7) % 100, 42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 11] SELECTION SORT
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 13] INSERTION SORT
def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
#[CELL 13] EXECUTION-TIME MEASUREMENT
from os import times
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10

performance_data = [
    (ID_SUM * 3 + i * 7) % 100
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort: {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```
